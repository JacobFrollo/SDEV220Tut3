```python
class Solution:
    # Function to sort an array of 0s, 1s, and 2s
    def sort012(arr):
        n0 = 0
        n1 = 0
        n2 = 0
    
        # Count 0s, 1s and 2s
        for num in arr:
            if num == 0:
                n0 += 1
            elif num == 1:
                n1 += 1
            else:
                n2 += 1
    
        index = 0
        
        # Place all the 0s
        for i in range(n0):
            arr[index] = 0
            index += 1
    
        # Place all the 1s
        for i in range(n1):
            arr[index] = 1
            index += 1
    
        # Place all the 2s
        for i in range(n2):
            arr[index] = 2
            index += 1
    
    
    arr = [0, 2, 2, 0, 1, 1]
    sort012(arr)
    
    for x in arr:
        print(x, end = " ")
```

```python
class Solution:

    def binarysearch(arr, k):
        left, right = 0, len(arr) - 1
        result = -1

        while left <= right:
            middle = left + (right - left) // 2

            if arr[middle] == k:
                result = middle
                right = middle - 1  # Search in the left half for the first occurrence
            elif arr[middle] < k:
                left = middle + 1
            else:
                right = middle - 1

        return result

    arr = [1, 2, 3, 4, 5]
    k = 3
    print(binarysearch(arr, k))
```
